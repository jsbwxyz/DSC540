### Exercise 2

Create a Jupyter notebook where you create a list, iterate over the list and sort your results, generate random numbers, add to the list, and then print your results.

```python
# Creating a list of fans
types_of_fans = ['ceiling','box','tower','desk','bladeless']
types_of_fans
```

```python
# Iterating over list of fans
def iterate_fans():
    for fan in types_of_fans:
        print(fan)
iterate_fans()
```

```python
# Alphabetically sorting the list of fans
types_of_fans.sort()
iterate_fans()
```

```python
# Add random numbers to add to the list of fans
# Random added if 's' is found
import random

# Create a function...
def add_random():
    try:
        for fan in types_of_fans:
            # ..to add a random number
            num = random.random()
            # if an 's' is found within the 
            #   letters of the strings in types_of_fans
            if 's' in fan:
                types_of_fans.append(num)
    except TypeError:
        # Function may pick up it's own entries
        #   which will throw an error
        print('Operation completed.')
            
add_random()
```

```python
# This, is now types_of_fans
types_of_fans
```

```python
# Bonus: Spin the list of fans

# Clear numerics
types_of_fans[:] = [x for x in types_of_fans if isinstance(x, str)]
#  targets list... keep x only if...  # ^^ this condition is true ^^

# Pass in the list as a variable
def spin(types_of_fans):

    # If the list exists, (always true)
    if types_of_fans:
        #   then move the list up one spot, .pop(0)
        types_of_fans.append(types_of_fans.pop(0))

spin(types_of_fans)

# Running excessively may cause slight breeze.
types_of_fans
```
