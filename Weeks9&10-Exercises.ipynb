Exercise 7.02: Checking the Status of the Web Request

Checking the status code and flag an error/no error situation. The 1 and -1 are utilized in the below function to flag if the request itself failed or not.

```python
# Creating the check status function
def status_check(r):
    if r.status_code==200:
        print("Success!")
        return 1
    else:
        print("Failed!")
        return -1
```

```python
response = requests.get(wiki_home)
status_check(response)
```
