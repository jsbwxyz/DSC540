#### Project Subject Area

This project analyzes the impact of government spending and financial health on the local retail job market. State IDs, also identified by codes, will be the primary keys of the database.

#### Data Sources

##### <b>API</b> 
USAspending.gov API<br>
Spending by Award Endpoint<br>

This data represents money awarded to states through private business. "Total Obligation" is the specific dollar amount that was paid by the government, and "Place of Performance" is the STATE code, also the primary key. These are the 'what' and 'where' of each row, which could be seen as an individual receipt. Collecting this data is part of how we understand the financial health of a state in our analysis.

https://api.usaspending.gov/api/v2/search/spending_by_award/


##### <b>Web Table</b> 
U.S Department of the Treasury <br>
Summary of Receipts and Outlays of the U.S. Government<br>

According to discaldatareasury.gov, the Monthly Treasury Statement (MTS) dataset provides information on the flow of money into and out of the U.S. Department of the Treasury. There is a categorized dataset of this table called the Summary of Receipients and Outlays of U.S Government which, by using the totals and Agency name to find STATE codes, which are the primary key.

https://fiscaldata.treasury.gov/datasets/monthly-treasury-statement/summary-of-receipts-outlays-and-the-deficit-surplus-of-the-u-s-government

##### <b>Flat File</b>
Bureau of Labor Statistics <br>
Quarterly Census of Employment and Wages<br>

QCEW provides a colection of CSV files representing logical slices of data on employment and wages from different slices. The industry slice is what will be used to collect data at an industrial level about employment and wages. FIPS codes are used by the government and correspond directly to States, which can be used as the primary key.

https://data.bls.gov/cew/data/api/2024/1/industry/10.csv
