#### Project Subject Area

This project analyzes the impact of government spending and financial health on the local retail job market. State IDs, also identified by codes, will be the primary keys of the database.

##### <b>API</b>

### Summary of Relationships

Essentially, this project relies on joining datasets across two distinct keys, Agency and Geography.

The Treasury data lists one total budget for an agency. In a One To Many relationship, the treasury website provides the high-level parent data ("Individual Contract Awards by Agency"). These will be linked by normalizing the Agency Names.
    
The BLS CSV file identifies states using numeric FIPS codes (e.g., 48), whereas the API uses Postal Codes (e.g., TX). Mapping will be used tocreated to link these two distinct identifiers, enabling the correlation of federal award dollars with state-level retail wage data.

<b>Project Plan & Approach</b>

The goal of this project is to understand if government spending actually helps raise weekly wages for people working in retail. To figure this out, data will be collected from three different places: a CSV file with state wage info, an API with federal contract details, and a website table showing total agency budgets.

The biggest technical step is fixing the fact that these files use different "codes" for states. The wage file uses numbers (FIPS codes), but the contract API uses letters (State Abbreviations). A big part of the work will be building a simple map to link these two together. This allows the data to be joined correctly so that federal spending can be compared to the actual wages in each state. 

By the end, the information will be combined into one large table with over 1,000 rows and 30 columns to show the real impact of federal spending on local retail paychecks.


<b>Concerns and Challenges</b>

The main concern is the data cleaning needed to get these three sources to talk to each other. Because the Treasury website, the BLS file, and the USAspending API are all run by different groups, they don't use the same names for agencies or the same ID numbers for states. If the "Department of Defense" is spelled differently in two files, the computer won't know they are the same thing.

Another challenge is timing. The government might sign a contract today, but the money might not hit the local economy for months. This "delay" can make it tricky to see a direct link between a specific award and a change in wages. Making sure the dates in the API match up with the dates in the wage CSV will be a top priority.

<b>Ethical Implications</b>

When looking at economic data, it is important to consider the privacy and livelihoods of the people involved. Even though this data is public, it represents the real-world income of workers. The analysis must be careful not to unfairly judge or misrepresent specific low-income areas or small businesses.

There is also a risk of bias when looking at the results. For example, if the data shows high spending but low wages, it might look like the spending "failed." However, that money might be helping the community in other ways, like better roads or national security. This project will focus on presenting the facts clearly without jumping to unfair conclusions.
