#### Activity 7.0.1: Extracting the top 100 E-Books from Gutenberg

1. Import the necessary libraries, including regex and BeautifulSoup.

```python
import urllib.request, urllib.parse, urllib.error
import requests
from bs4 import BeautifulSoup
import ssl
import re
```

2. Read the HTML from the URL:

```python
toponehun = 'https://www.gutenberg.org/browse/scores/top'
response = requests.get(toponehun)
```

3. Write a small function to check the status of a the web request.

```python
def status_check(r):
    if r.status_code==200:
        print("Success!")
        return 1
    else:
        print("Fialed!")
        return -1
```

```python
# Check response
status_check(response)
```

4. Decode the response and pass it on to BeautifulSoup for HTML parsing.

```python
contents = response.content.decode(response.encoding)
soup = BeautifulSoup(contents, 'html.parser')
soup
```

5. Find all the href tags and store them in a list of links.

```python
links = []
# Store all href tags, store in list of links
for link in soup.find_all('a'):
    print(link.get('href'))
    links.append(link.get('href'))
```
