Exercise 3.19: Generating Random Numbers from a Binomial Distribution and
Bar Plot

```python
# Import the numpy library
import numpy as np
```

```python
# Let's generate random numbers from a binomial distribution. This experiment
x = np.random.binomial(10,0.6,size=8)
print(x)
```

```python
# Plot the result using a bar chart
import matplotlib.pyplot as plt
plt.figure(figsize=(7,4))
plt.title("Number of success in coin toss", fontsize=16)
plt.bar(np.arange(1,9), height=x)
plt.xlabel("Exp. number", fontsize=15)
plt.ylabel("Number of successes", fontsize=15)
plt.show()
```

Exercise 3.20: Generating Random Numbers from a Normal Distribution and
Histograms

```python
import numpy as np
```

```python
# Draw a single sample from a normal distribution by using the
#   following command:
x = np.random.normal()
print(x)
```

```python
# Generate a histogram of 100 students by using the following command
import matplotlib.pyplot as plt
# Code to generate the 100 samples (heights)
heights = np.random.normal(loc=155,scale=10,size=100)
```

```python
# Plotting code
plt.figure(figsize=(7,5))
plt.hist(heights,color='orange',edgecolor='k')
plt.title("Histogram of teenaged students' height",\
fontsize=18)
plt.xlabel("Height in cm",fontsize=15)
plt.xticks(fontsize=15)
plt.yticks(fontsize=15)
plt.show()
```

Exercise 3.21: Calculating Descriptive Statistics from a DataFrame

```python
import numpy as np
import pandas as pd
people = ['Ann','Brandon','Chen',\
'David','Emily','Farook',\
'Gagan','Hamish','Imran',\
'Joseph','Katherine','Lily']
age = [21,12,32,45,37,18,28,52,5,40,48,15]
weight = [55,35,77,68,70,60,72,69,18,65,82,48]
height = [160,135,170,165,173,168,175,159,105,171,155,158]
people_dict={'People':people,'Age':age,\
'Weight':weight,'Height':height}
people_df=pd.DataFrame(data=people_dict)
people_df
```

```python
# Find the number of rows and columns of the DataFrame by executing the
#   following command:
print(people_df.shape)
```

```python
# Obtain a simple count (any column can be used for this purpose) by executing
#   the following command:
print(people_df['Age'].count())
```

```python
# Calculate the mean age by using the following command:
print(people_df['Age'].mean())
```

```python
# Calculate the maximum height by using the following command:
print(people_df['Height'].max())
```

```python
# Calculate the standard deviation of the weights by using the following command:
print(people_df['Weight'].std())
```

```python
# Calculating percentile
pcnt_75 = np.percentile(people_df['Age'], 75)
pcnt_25 = np.percentile(people_df['Age'],25)
print("Inter-quartile range: ",pcnt_75-pcnt_25)
```

```python
# Use the describe command to find a detailed description of the DataFrame:
print(people_df.describe())
```
